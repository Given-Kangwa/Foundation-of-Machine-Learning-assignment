```python
#Pruning of weights from the data set 
import numpy as np 
```

```python
mlp_full = MLPRegressor(hidden_layer_sizes=(64,32,16),random_state=42,max_iter=1000)
```

```python
mlp_full_pipeline = make_pipeline(SimpleImputer(strategy='mean'),mlp_full)

mlp_full_pipeline.fit(X_train,y_train)
```

```python
y_full = mlp_full_pipeline.predict(X_test)
```

```python
weights = mlp_full_pipeline.named_steps['mlpregressor'].coefs_
```

```python
print(weights)
```

```python
# simple pruning: zero small weights
pruned_weights = []
threshold = 0.01
```

```python
for w in weights:
    w_pruned = np.where(np.abs(w) < threshold, 0, w)
    pruned_weights.append(w_pruned)
```

```python
mlp_full_pipeline.coefs_ = pruned_weights
```

```python
# correct place for the pruned weights
mlp_full_pipeline.named_steps['mlpregressor'].coefs_ = pruned_weights

# predict with the pipeline
y_pruned = mlp_full_pipeline.predict(X_test)
```

```python
print("MSE (Pruned MLP):",mean_squared_error(y_test, y_pruned))
print("R² (Pruned MLP):",r2_score(y_test, y_pruned))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

standard = StandardScaler()

pipeline_ =make_pipeline(SimpleImputer(strategy = 'mean'),standard) 

# Keep only numeric features for imputation/scaling
X_train = X_train.select_dtypes(include=['number'])
X_test = X_test.select_dtypes(include=['number'])

X_st = pipeline_.fit_transform(X_train)
X_test_st = pipeline_.transform(X_test)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=0.85)

X_train_pca = pca.fit_transform(X_st)
X_test_pca = pca.transform(X_test_st)
```

```python
from sklearn.cluster import KMeans
from sklearn.linear_model import LinearRegression
import numpy as np

# 1. Generate cluster IDs as features
kmeans = KMeans(n_clusters=3, random_state=42)
train_clusters = kmeans.fit_predict(X_train_pca).reshape(-1, 1)
test_clusters = kmeans.predict(X_test_pca).reshape(-1, 1)

# 2. Combine your PCA features with the new cluster feature
X_train_combined = np.hstack((X_train_pca, train_clusters))
X_test_combined = np.hstack((X_test_pca, test_clusters))

# 3. Train a regression model on the combined data
regressor = LinearRegression()
regressor.fit(X_train_combined, y_train)

# 4. Predict and evaluate
y_pred = regressor.predict(X_test_combined)
print('Mean_squared_error = ', mean_squared_error(y_test, y_pred))
print('r2_score = ', r2_score(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score

KNN = KNeighborsRegressor(n_neighbors = 4)

KNN.fit(X_train_pca,y_train)

y_pred_knn = KNN.predict(X_test_pca)

print('mean_squared_error for KNN =',mean_squared_error(y_test,y_pred_knn))
print('r2_score for KNN =', r2_score(y_test,y_pred_knn))
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

tsne = TSNE(n_components=2, random_state=42, perplexity=30)
X_tsne = tsne.fit_transform(X_train_pca)

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=train_clusters.ravel(), cmap='viridis', alpha=0.6)
plt.title('t-SNE projection of PCA features')
plt.xlabel('t-SNE 1')
plt.ylabel('t-SNE 2')
plt.show()
```
